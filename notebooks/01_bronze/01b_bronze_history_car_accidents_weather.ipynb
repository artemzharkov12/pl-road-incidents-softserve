```python
from pyspark.sql.types import StructType, StructField, StringType
```

```python
dbutils.widgets.text("bronze_catalog", "dbr_dev")
dbutils.widgets.text("bronze_schema", "artemzharkov10_bronze")
dbutils.widgets.text("raw_data_folder", "weather_year")

BRONZE_CATALOG = dbutils.widgets.get("bronze_catalog")
BRONZE_SCHEMA = dbutils.widgets.get("bronze_schema")
RAW_DATA_FOLDER = dbutils.widgets.get("raw_data_folder")
```

```python
VOLUME_PATH = f"/Volumes/{BRONZE_CATALOG}/{BRONZE_SCHEMA}/raw_data/{RAW_DATA_FOLDER}"
TABLE_NAME = f"{BRONZE_CATALOG}.{BRONZE_SCHEMA}.bronze_historical_car_accidents_weather"
```

```python
weather_schema = StructType([
    StructField("lat", StringType(), False),
    StructField("lon", StringType(), False),
    StructField("target_acc_date", StringType(), False),   
    StructField("time", StringType(), False),
    StructField("temperature_2m", StringType(), False),
    StructField("precipitation", StringType(), False),
    StructField("wind_speed_10m", StringType(), False),
    StructField("soil_temp", StringType(), False),
    StructField("humidity_2m", StringType(), False),
    StructField("dew_point_2m", StringType(), False)
])

df_history_car_accidents_weather = (
    spark.read
    .format("csv")
    .schema(weather_schema)
    .option("header", "true")
    .load(VOLUME_PATH)
)

(df_history_car_accidents_weather.write
    .mode("overwrite")
    .format("delta")
    .saveAsTable(f"{TABLE_NAME}")
)
```
