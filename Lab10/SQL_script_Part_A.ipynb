Part A (join an external table with a Delta table)

```python
%sql
WITH accidents AS (
    SELECT
        *,
        CAST(FLOOR(rand() * 100) AS INT) AS random_driver_id
    FROM dbr_dev.artemzharkov10_silver.silver_streaming_road_accidents)
SELECT 
    accidents.accident_id,ext.driver_id,ext.first_name,ext.last_name,accidents.icon_category,accidents.start_time
FROM lab10_catalog.public.external_drivers ext
JOIN accidents
    ON ext.driver_id = accidents.random_driver_id;
```
